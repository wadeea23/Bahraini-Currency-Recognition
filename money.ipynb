```python
from ultralytics import YOLO
```

```python
model = YOLO("best.pt")
```

```python
import cv2

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    results = model(frame) # prediction
    annotated_frame = results[0].plot()  # draw boxes

    cv2.imshow("YOLO Detection", annotated_frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
import streamlit as st
from PIL import Image
from ultralytics import YOLO
from streamlit_webrtc import webrtc_streamer
import av

st.title("Bahraini Currency Recognition")

# Load model
model = YOLO("best.pt")

# Upload Image
st.header("Upload Image")

file = st.file_uploader("Choose an image", type=["jpg", "png", "jpeg"])

if file:
    img = Image.open(file).convert("RGB")

    result = model(img)[0]

    st.image(result.plot(), caption="Prediction")


# Live Camera
st.header("Live Camera")

def camera(frame):
    img = frame.to_ndarray(format="bgr24")

    result = model(img)[0]

    return av.VideoFrame.from_ndarray(result.plot(), format="bgr24")


webrtc_streamer(
    key="camera",
    video_frame_callback=camera
)
```
